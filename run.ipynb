# Summary data from group #8

```python
my_group = {
    "name": "vfdv52-Gumanji530",
    "members": [
        {"name": "Jill",
         "age": 26,
         "job": "biologist",
         "connections":{ 
         "friend": ["Zalika"],
         "partner": ["John"]}
         },
        {"name": "Zalika",
         "age": 28,
         "job": "Designer",
         "connections": {
         "friend": ["Jill"]}
         },
        {"name": "John",
         "age": 27,
         "job": "Writer",
         "connections": {
         "partner": ["Jill"]}
         },
        {"name": "Nash",
         "age": 34,
         "job": "Chef",
         "connections": {
         "cousin": ["John"],
         "landlord": ["Zalika"]}
        }
    ]
}
```

### The maximum age of people in the group

```python
def max_age(name):
    ages = [member["age"] for member in my_group["members"] if member["name"] == name]
    return max(ages) if ages else None
maximum_age = max_age("Alice")
print(f"The maximum age of members named Alice is: {maximum_age}")

maximum_age = max_age("John")
print(f"The maximum age of members named John is: {maximum_age}")
```

### The average (mean) number of relations among members of the group

```python
def average_relations():
    total_relations = 0
    for member in my_group["members"]:
        total_relations += sum(len(relations) for relations in member["connections"].values())
    average = total_relations / len(my_group["members"]) if my_group["members"] else 0
    return average

avg_relations = average_relations()
print(f"The average number of relations among members of the group is: {avg_relations}")
```

### The maximum age of people in the group that have at least one relation

```python
def eldest_with_relations():
    ages_with_relations = [member["age"] for member in my_group["members"] if any(member["connections"].values())]
    return my_group["members"] if ages_with_relations else None

eldest_age = eldest_with_relations()
print(f"The eldest age among members with relations is: {eldest_age}")
```

### [more advanced] The maximum age of people in the group that have at least one friend

```python
def eldest_age_with_friends():
    ages_with_friends = [member["age"] for member in my_group["members"] if member["connections"].get("friend")]
    return max(ages_with_friends) if ages_with_friends else None

eldest_age = eldest_age_with_friends()
print(f"The eldest age among members with friends is: {eldest_age}")
```
